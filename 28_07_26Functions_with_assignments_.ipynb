````python
def outer():

    x = 10

    def inner():
        nonlocal x
        x = 20

    inner()

    print(x)

outer()
````

````python
x = 100      # Global variable
def outer():
    y = 200   # Enclosing variable
    def inner():
        global x
        nonlocal y
        x = x + 50
        y = y + 50
        print("Inside Inner")
        print("Global x :", x)
        print("Enclosing y :", y)
    inner()
    print("Inside Outer")
    print("y =", y)
outer()
print("Outside")
print("x =", x)
````

````python
count = 0
def company():
    employees = 10
    def hire():
        global count
        nonlocal employees
        count += 1
        employees += 5
        print("Employees:", employees)
        print("Global Count:", count)
    hire()
    print("Final Employees:", employees)
company()
print("Total Companies Processed:", count)
````

````python
bank_balance = 10000
def account():
    savings = 5000
    def deposit():
        global bank_balance
        nonlocal savings
        bank_balance += 2000
        savings += 1000
        print("Savings =", savings)
        print("Bank Balance =", bank_balance)
    deposit()
    print("Account Savings =", savings)
account()
print("Global Balance =", bank_balance)
````

````python
x = 10
def outer():
    y = 20
    def middle():
        z = 30
        def inner():
            global x
            nonlocal y
            nonlocal z
            x += 1
            y += 2
            z += 3
            print("Inner")
            print(x, y, z)
        inner()
        print("Middle:", z)
    middle()
    print("Outer:", y)
outer()
print("Global:", x)
````

````python
### global : varaibles declared outside all functions : globa
### nonlocal: varaible declared in the nearest to the enclosing functions, nonlocal
### LEG : local, enclosing brackets, global
````

````python
#Assignment 92: Closure-Based Bank Balance
#Write a function named create_bank_account().
#Requirements
#Store the balance in the outer function.
#Define inner functions for:
#Deposit
#Withdrawal
#Balance enquiry
#Use nonlocal to update the balance.
#Reject withdrawals greater than the available balance.
#Return all three inner functions.
#Hint
#Return multiple functions:
#return deposit, withdraw, check_balance
def create_bank_account(initial_balance):
    balance = initial_balance

    def deposit(amount):
        nonlocal balance
        balance += amount
        return balance

    def withdraw(amount):
        nonlocal balance

        if amount > balance:
            return "Insufficient balance"

        balance -= amount
        return balance

    def check_balance():
        return balance

    return deposit, withdraw, check_balance


deposit, withdraw, check_balance = create_bank_account(1000)

print("Initial Balance:", check_balance())

print("After Deposit:", deposit(500))

print("After Withdrawal:", withdraw(300))

print("Current Balance:", check_balance())
````

````python
#Assignment 93: Function Decorator
#Create a decorator named execution_logger().
#Requirements
#he decorator should print a message before the original function executes.
#It should print another message after execution.
#Apply it to an addition function.
#Preserve the original function's return value.
#xpected Pattern
#Starting function...
#Result: 30
#Function completed
def execution_logger(function):

    def wrapper(*args, **kwargs):
        print("Starting function...")

        result = function(*args, **kwargs)

        print("Result:", result)
        print("Function completed")

        return result

    return wrapper


@execution_logger
def add(a, b):
    return a + b


result = add(10, 20)
````

````python
#Assignment 100: Complete Functional Programming Project

#Develop a complete Employee Performance Analysis System using functions, lambda expressions, map(), filter(), reduce(), nested functions, closures, and LEGB scope.

#Sample Employee Data
#employees = [
   # {
       # "id": 101,
       # "name": "Ravi",
       # "department": "IT",
       # "salary": 60000,
       # "performance_score": 88,
       # "experience": 5
    #},
    #{
       # "id": 102,
       # "name": "Anitha",
        #"department": "HR",
        #"salary": 45000,
        #"performance_score": 72,
        #"experience": 4
    #},
    #{
       # "id": 103,
       # "name": "Kiran",
        #"department": "IT",
        #"salary": 75000,
        #"performance_score": 94,
        #"experience": 7
    #},
    #{
       # "id": 104,
        #"name": "Meena",
        #"department": "Finance",
        #"salary": 55000,
        #performance_score": 81,
        #"experience": 6
    #},
    #{
       # "id": 105,
       # "name": "Arjun",
       # "department": "IT",
        #"salary": 40000,
        #"performance_score": 65,
        #"experience": 2
    #}
#]


def generate_department_report(department):

    department_employees = list(
        filter(
            lambda employee:
            employee["department"].lower() == department.lower(),
            employees
        )
    )

    def count_employees():
        return len(department_employees)

    def total_salary():
        return reduce(
            lambda total, employee:
            total + employee["salary"],
            department_employees,
            0
        )

    def average_salary():
        return total_salary() / count_employees()

    def average_performance():
        total = reduce(
            lambda total, employee:
            total + employee["performance_score"],
            department_employees,
            0
        )

        return total / count_employees()

    return {
        "department": department,
        "employee_count": count_employees(),
        "total_salary": total_salary(),
        "average_salary": round(average_salary(), 2),
        "average_performance": round(average_performance(), 2)
    }

    report = generate_department_report("IT")
print(report)
````
